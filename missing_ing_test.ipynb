```python
# Parameters
def test_proportion(proportion=0):
    num_samples = 50
    ing_range = (8, 16)
    transcript_ings = 4
    num_examples = 16
    transcript_length = 4
    spec=0

    print(f'Testing {proportion * 16} irrelevant options:')

    result = test(num_samples, ing_range, transcript_ings, num_examples, transcript_length, spec, proportion=proportion)

    outputs, full_outputs, metadata = result

    dataroot = f'./results/t{transcript_ings}_n{num_examples}_l{transcript_length}_s{spec}_({ing_range[0]},{ing_range[1]})_p{int((1-proportion)*num_examples)}'

    try: os.mkdir(dataroot)
    except OSError as e: pass

    meta_df = pd.DataFrame(metadata)
    out_df = pd.DataFrame(outputs)
    full_df = pd.DataFrame(full_outputs)

    meta_df.to_csv(f'{dataroot}/metadata.csv', index=False)
    out_df.to_csv(f'{dataroot}/outputs.csv', index=False)
    full_df.to_csv(f'{dataroot}/full_outputs.csv', index=False)
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt

def get_accuracies(dir):
    accuracies = {}
    for model in openai_list: accuracies[model] = 0
    for model in together_list: accuracies[model] = 0

    metadata = pd.read_csv(f'{dir}/metadata.csv')
    outputs = pd.read_csv(f'{dir}/outputs.csv')
    for col in outputs.columns:
        raw = [text for text in outputs.loc[:, col]]
        correct = [(r.rfind(metadata['correct'][i]) != -1 and
                    r.rfind('Recipe') == r.rfind(metadata['correct'][i]))
                   for i, r in enumerate(raw)]

        mean = len(outputs.loc[:, col][correct]) / len(raw)
        sd = math.sqrt((mean * (1-mean)) / len(raw))
        accuracies[col] = mean, sd

    return accuracies

def plot_comp(dir1, dir2, labels):
    acc1 = get_accuracies(dir1)
    acc2 = get_accuracies(dir2)

    means = {}
    sds = {}
    for label in labels:
        means[label] = []
        sds[label] = []

    for model in acc1:
        mean1, sd1 = acc1[model]
        mean2, sd2 = acc2[model]
        means[labels[0]].append(mean1)
        sds[labels[0]].append(sd1)
        means[labels[1]].append(mean2)
        sds[labels[1]].append(sd2)


    x = np.arange(len(acc1))  # the label locations
    width = 0.25  # the width of the bars
    multiplier = 0

    fig, ax = plt.subplots(layout='constrained')

    for label in means:
        offset = width * multiplier
        ax.bar(x + offset, means[label], width, edgecolor='black', label=label)
        ax.errorbar(x + offset, means[label], yerr = sds[label], fmt ='o', color='black')
        multiplier += 1

    ax.set_xticks(range(len(acc1)), [model.split('/')[-1] for model in acc1.keys()], rotation=30)
    ax.set_ylim(0,1)

    plt.title(f'{labels[0]} vs. {labels[1]}')
    plt.legend()
    plt.show()

def plot_line_logx(ex:list, ambiguous=False):
    means = {}
    sds = {}
    for model in openai_list:
        means[model] = []
        sds[model] = []
    for model in together_list:
        means[model] = []
        sds[model] = []

    xticks = [int(d.split('/')[-1].split('_')[1][1:]) for d in ex]
    xticks.sort()

    for i in np.argsort([int(d.split('/')[-1].split('_')[1][1:]) for d in ex]):
        metadata = pd.read_csv(f'{ex[i]}/metadata.csv')
        outputs = pd.read_csv(f'{ex[i]}/outputs.csv')
        for col in outputs.columns:
            if col not in openai_list and col not in together_list: continue
            raw = [text for text in outputs.loc[:, col]]
            correct = [r.split('<answer>')[-1].split('</answer>')[0] != metadata['correct'][j]
                       if ambiguous else
                            (r.rfind(metadata['correct'][j]) != -1 and
                             r.rfind('Recipe') == r.rfind(metadata['correct'][j]))
                       for j, r in enumerate(raw)]

            mean = len(outputs.loc[:, col][correct]) / len(raw)
            sd = math.sqrt((mean * (1-mean)) / len(raw))
            # accuracies[col] = mean, sd
            means[col].append(mean)
            sds[col].append(sd)

    # print(means)
    # print(sds)

    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(layout='constrained', figsize=(15, 8))

    for i, model in enumerate(means):
        print(model)
        mean = np.array(means[model])
        sd = np.array(sds[model])
        ax.plot(xticks, mean, label=model.split('/')[-1])
        # ax.errorbar(range(len(mean)), mean, yerr = sd, fmt ='o', color='black')
        ax.fill_between(xticks, mean-sd, mean+sd ,alpha=0.3)

    control = np.arange(1, xticks[-1], step=0.1)
    if ambiguous: ax.plot(control, 1 - 1/control, '--', label='Baseline (Random choice)')
    else: ax.plot(control, 1/control, '--', label='Baseline (Random choice)')

    ax.set_xlim(2,128)
    ax.set_ylim(0,1)
    ax.set_title('Results')
    ax.set_xscale('log')
    ax.set_xticks(xticks, [f'{x} options' for x in xticks], rotation=30)
    ax.grid(True)

    fig.legend(loc='outside right upper')
    plt.show()


def plot_line(ex:list):
    means = {}
    sds = {}
    for model in openai_list:
        means[model] = []
        sds[model] = []
    for model in together_list:
        means[model] = []
        sds[model] = []

    xticks = [float(d.split('_p')[1]) for d in ex]
    xticks.sort()

    for i, folder in enumerate(ex):
        metadata = pd.read_csv(f'{ex[i]}/metadata.csv')
        outputs = pd.read_csv(f'{ex[i]}/outputs.csv')
        for col in outputs.columns:
            raw = [t for t in outputs.loc[:, col]]
            correct = [(str(r).rfind(metadata['correct'][j]) != -1 and
                    str(r).rfind('Recipe') == str(r).rfind(metadata['correct'][j]))
                   for j, r in enumerate(raw)]

            mean = len(outputs.loc[:, col][correct]) / len(raw)
            sd = math.sqrt((mean * (1-mean)) / len(raw))
            # accuracies[col] = mean, sd
            means[col].append(mean)
            sds[col].append(sd)

    # print(means)
    # print(sds)

    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(layout='constrained')

    for i, model in enumerate(means):
        mean = np.array(means[model])
        sd = np.array(sds[model])
        ax.plot(xticks, mean, label=model.split('/')[-1])
        # ax.errorbar(range(len(mean)), mean, yerr = sd, fmt ='o', color='black')
        ax.fill_between(xticks, mean-sd, mean+sd ,alpha=0.3)

    ax.set_xlim(1,16)
    ax.set_ylim(0,1)
    ax.set_title('Results')
    ax.set_xscale('log')
    ax.set_xticks(xticks, [f'{int(x)}/16 relevant option(s)' for x in xticks], rotation=30)
    ax.grid(True)

    fig.legend(loc='outside right upper')
    plt.show()


def plot_comp(ex1, ex2, xticks, labels):
    def extract(ex):
        means = {}
        sds = {}
        for model in openai_list:
            means[model] = []
            sds[model] = []
        for model in together_list:
            means[model] = []
            sds[model] = []

        for i in np.argsort([int(d.split('_')[1][1:]) for d in ex]):
            metadata = pd.read_csv(f'{ex[i]}/metadata.csv')
            outputs = pd.read_csv(f'{ex[i]}/outputs.csv')
            for col in outputs.columns:
                raw = [text for text in outputs.loc[:, col]]
                correct = [(r.rfind(metadata['correct'][j]) != -1 and
                        r.rfind('Recipe') == r.rfind(metadata['correct'][j]))
                       for j, r in enumerate(raw)]

                mean = len(outputs.loc[:, col][correct]) / len(raw)
                sd = math.sqrt((mean * (1-mean)) / len(raw))
                # accuracies[col] = mean, sd
                means[col].append(mean)
                sds[col].append(sd)
        return means, sds

    m1,s1 = extract(ex1)
    m2,s2 = extract(ex2)

    # print(f'{m1} vs. {m2}')


    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(max(len(openai_list), len(together_list)), 2, layout='tight', figsize=(15, 15))

    def plot(models, ind):
        for i in range(len(ax)):
            if i >= len(models):
                fig.delaxes(ax[i][ind])
                continue

            model = models[i]
            mean1 = np.array(m1[model])
            sd1 = np.array(s1[model])
            mean2 = np.array(m2[model])
            sd2 = np.array(s2[model])

            ax[i][ind].plot(xticks, mean1, label=labels[0])
            ax[i][ind].fill_between(xticks, mean1-sd1, mean1+sd1 ,alpha=0.3)

            ax[i][ind].plot(xticks, mean2, label=labels[1])
            ax[i][ind].fill_between(xticks, mean2-sd2, mean2+sd2 ,alpha=0.3)

            ax[i][ind].set_xlim(min(xticks),max(xticks))
            ax[i][ind].set_ylim(0,1)
            ax[i][ind].set_title(model)
            ax[i][ind].set_xscale('log')
            ax[i][ind].set_xticks(xticks, [f'{int(x)} relevant option(s)' for x in xticks], rotation=30)
            ax[i][ind].grid(True)

            control = np.arange(1, xticks[-1], step=0.1)
            ax[i][ind].plot(control, 1/control, '--', label='Baseline (Random choice)')

    plot(openai_list, 0)
    plot(together_list, 1)

    handles, labels = ax[0,0].get_legend_handles_labels()
    fig.legend(handles, labels, loc='upper center')
    # plt.figure(figsize=(15, 8)).suptitle(f'Effect of Irrelevant Option Padding', fontsize=20)
    # fig.legend(loc='outside right upper')
    # plt.tight_layout()
    plt.show()
```

```python
dataroot = 'results/t4_n16_l4_s0_(8,16)_p4'
model = 'Qwen/Qwen3.5-397B-A17B'

meta_df = pd.read_csv(f'{dataroot}/metadata.csv')
out_df = pd.read_csv(f'{dataroot}/outputs.csv')
full_df = pd.read_csv(f'{dataroot}/full_outputs.csv')

qwen_col = out_df.loc[:, model]

timeout_inds = [i for i, out in enumerate(qwen_col) if out == 'timeout']
for i in tqdm(timeout_inds):
    prompt = meta_df.loc[i, 'full_prompt']
    client = Together()
    try:
        response = client.chat.completions.create(
            model=model,
            messages=[
                # {"role": "user", "content": instructions},
                # {"role": "user", "content": instructions + transcript_text + mistake_text}
                {"role": "user", "content": prompt}
            ],
            reasoning={"enabled": True},
            max_tokens=30000,
            timeout=300
        )
        out_df.loc[i, model] = response.choices[0].message.content
        full_df.loc[i, model] = response
    except Exception as e: print(f'Error: {str(e)} ({model})')

out_df.to_csv(f'{dataroot}/outputs.csv', index=False)
full_df.to_csv(f'{dataroot}/full_outputs.csv', index=False)
```
